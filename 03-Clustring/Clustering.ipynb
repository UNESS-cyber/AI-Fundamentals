```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn import cluster
from sklearn import datasets
from sklearn.preprocessing import MinMaxScaler,StandardScaler
from sklearn.cluster import KMeans
from sklearn.cluster import DBSCAN
```

```python
nSamples = 500
seed = 30
noisyCircles = datasets.make_circles(n_samples=nSamples, factor=.5, noise=.05, random_state=seed)
noisyMoons = datasets.make_moons(n_samples=nSamples, noise=.05, random_state=seed)
blobs = datasets.make_blobs(n_samples=nSamples, random_state=seed)
rng = np.random.RandomState(seed)
noStructure = rng.rand(nSamples, 2), np.zeros(nSamples, np.int8)
randomState = 170
x , y = datasets.make_blobs(n_samples=nSamples, random_state=randomState)
transformation = [[0.6,-0.6],[-0.4,0.8]]
Xaniso = np.dot(x, transformation)
aniso = (Xaniso, y)
varied = datasets.make_blobs(n_samples = nSamples, cluster_std = [1.1, 2.5, 0.7] , random_state=randomState)
```

```python
fig , ax = plt.subplots(5 , 1 , figsize = (5,15))
color = np.array(["Blue","Red","Green"])
for (i,datasets) in enumerate([blobs , noisyMoons , noStructure , aniso , varied]):
    ax[i].scatter(datasets[0][:,0], datasets[0][:,1], c=color[datasets[1]])
```

## Preprocessing

```python
dataset = blobs
x , y = dataset
x = MinMaxScaler().fit_transform(x)
plt.scatter(x[:, 0], x[:, 1], alpha=1.0, color='b')
```

## Clustering
### **Within-Cluster Sum of Squares (WCSS)** metric

```python
wcss = []
model = []
for i in range(1, 11):
    model.append(KMeans(n_clusters=i).fit(x))
    wcss.append(model[i-1].inertia_)
```

```python
plt.plot(range(1, 11), wcss)
plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
yp = model[2].fit_predict(x)
cmap = plt.cm.jet
fig , ax = plt.subplots()
ax.scatter(x[:, 0], x[:, 1], c=yp, cmap=cmap)
```

## DBSCAN
### **Density-Based Spatial Clustering Applications With Noise**

```python
model = DBSCAN(eps=0.05, min_samples=10)
yp = model.fit_predict(x)
fig , ax  = plt.subplots()
ax.scatter(x[:, 0], x[:, 1], c=yp + 1, cmap=cmap)
```

## Conclusion
### As demonstrated by the comparative results across all datasets, both the KMeans and DBSCAN algorithms are well-suited for clustering unlabeled data. A notable observation is that KMeans performs better on datasets with circular distributions, while DBSCAN is more effective for datasets with linear density patterns. The clustering tasks were carried out using the code developed during class sessions, applied consistently across various datasets.
### An important insight emerged when working with the "noisy circle" dataset: DBSCAN yielded superior results. This is because, contrary to expectations, KMeans struggles to correctly cluster data with hollow circular structures, often misclassifying points due to its reliance on centroid-based partitioning.
### In a separate file, an improved method for identifying more suitable clusters was implemented. This included defining reusable functions to streamline similar tasks and facilitate analysis. It is worth noting that this alternative approach is likely to produce even better clustering outcomes, especially when dealing with complex or irregular data distributions.
